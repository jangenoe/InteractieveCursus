### Notes — Slide 8

Intern bestaat de opamp uit een aantal verschilversterkers in serie achter elkaar geplaatst. Het effectieve design van deze verschilversterkers wordt uitgewerkt in een volgende cursus. In dit vak wil ik kort even een idee geven van waaruit deze verschilversterker opgebouwd is.
De  ingangssignalen v+ en v- komen toe bij v en v. Dit spanningsverschil levert een stroomverschil op door beide pMOS transistors in serie met deze nMOS transistors. Deze beide stromen worden gespiegeld in stromen die B maal groter zijn door de buitenste takken. De stroom door de linker buitenste tak wordt nog eens gespiegeld naar de rechter tak. De uitgang is aangesloten aan de rechter tak.
Als de stroom door de transistor met ingang V groter is, gaat $V_{out}$ dus snel omhoog en in het andere geval sterk omlaag.
Sinds de jaren  worden OPAMPs met MOS transistors gemaakt. De oude modellen (type ) met bipolaire transistors blijven hier en daar nog in gebruik

### Notes — Slide 33

In de eerste schakelingen gaan we gebruik maken van de diode als niet-lineair element. De diode heeft, in het geval van een voorwaarts aangelegde spanning, een stroom die exponentieel afhankelijk is van de aangelegde spanning.
Het is echter zeer belangrijk te benadrukken dat deze exponentiele karakteristiek slechts bekomen wordt in een zeer beperkt spanninggebied:
Wanneer de spanning te groot is treedt hoge injectie op, en hierbij wordt er steeds meer afgeweken van de zuivere logaritmische karakteristiek. We merken ook op dat een eventueel aanwezige serieweerstand hier een belangrijke afwijking geeft.
Wanneer de spanning te klein is, komt men in het gebied van lage injectie. Ook hier is er een afwijking van het zuiver exponentieel gedrag.
Bij het gebruik van diodes in de schakelingen die verder volgen is het dus belangrijk erover te waken dat de diode steeds in zijn normale werkingsgebied blijft. Ook de keuze van de diode is zeer belangrijk. Voor deze schakelingen heb je diodes van een goede kwaliteit nodig.

### Notes — Slide 57

Voor we de schakeling bespreken die een absolute waarde uitrekent bestuderen we eerst een hulpschakeling waarvan de karakteristiek aan de rechterzijde is weergegeven.
Indien $V_{in}$ positief is, wordt de uitgang van de OPAMP negatief. Er is geen terugkoppeling zolang er geen . volt voorwaartse spanning over de onderste diode staat. Van zodra dit echter het geval is, daalt Vhulp, waardoor er stroom loopt door de weerstanden. Er is negatieve terugkoppeling, waardoor de negatieve ingang nul wordt.
Indien $V_{in}$ negatief is, wordt de uitgang van de OPAMP positief. Er is geen terugkoppeling zolang er geen . volt voorwaartse spanning over de bovenste diode staat. Van zodra dit echter het geval is, wordt de negatieve ingang nul. Hierdoor wordt de uitgang ook nul.
Wanneer we beide diodes omkeren, inverteert deze schakeling de negatieve ingangsspanningen en wordt de uitgang nul in het geval van een positieve ingangsspanning.
Deze hulpschakeling heeft nog heel wat andere toepassingen. Zo laat deze schakeling toe van eender welke gevraagde karakteristiek te benaderen door een stapsgewijze lineaire functie.
