### Notes — Slide 9

Welke materialen komen in aanmerking als halfgeleider om Si  (silicium) te vervangen als halfgeleider voor vermogencomponenten. In eerst instantie elementen van kolom IV en combinaties van elementen van kolom III met kolom V. En de elementen die hoger in de tabel gelegen zijn leveren betere prestaties voor hoge vermogen. Het is echter zeer moeilijk (en dus duur) om een perfecte diamant (C) vandaar dat men naar SiC kan kijken als ergens tussenin. Als alternatief kan men ook kijken naar III-V combinaties zoals  en AlN. Het is pas de laatste jaren dat men er in slaagt deze materialen van hoge kwaliteit te maken voor vermogentoepassingen.
