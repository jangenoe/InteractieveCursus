## Slide 7
Current in the linear region
If $V_{DS}$ is small, the behavior is resistive


### Notes — Slide 7

The graph above expresses the source-drain current as a function of drain-source voltage and this for different values of $V_{GS}$. Whenever $V_{GS}$ is smaller than $V_T$, almost no current will flow, which corresponds to the zero line. When $V_{GS}$ gets larger, we obtain the subsequent lines on the graph.
In the current voltage characteristics of the MOS transistor we can differentiate between two domains:
A linear domain where at small $V_{DS}$ voltages the current linear increases with $V_{DS}$. $V_{GS}$-$V_T$ determines the conductivity of the component. When the $V_{DS}$ voltage further increases, the quadratic term in the current gains importance.
From a certain $V_{DS}$ voltage, the curve get perfectly flat. From that moment on the equation for the linear regime losses its validity and the current remains constant when the $V_{DS}$ voltage further increases.


## Slide 8
Border between linear and saturation
As long as the source-drain spanning is smaller than the saturation voltage, the device is in the linear region
At a higher source-drain voltage, the device reaches the saturation region.


### Notes — Slide 8

The transition between the linear and the saturation regime can easily be determined form the equation in the linear region. We take the derivative of this equation with respect to $V_{DS}$ and put this derivative equal to zero.


## Slide 9
Current in the saturation region
Current is independent of the Drain-Source voltage
Current evolves quadratic with Gate-Source voltage
The bulk voltage causes a shift of the $V_T$


### Notes — Slide 9

The equation for the current in the saturation regime can be obtained by entering the $V_{DS}$ value that defines the border line between the linear and saturation regime in the equation of the linear regime. The obtained equation is independent of $V_{DS}$. Therefor it is favorable to draw the current in the saturation regime as a function of $V_{GS}$ rather than as a function of $V_{DS}$. This curve is the transfer curve.
We note that this curve is valid for $V_{BS}$=0. When the bulk-source voltage ($V_{BS}$) is different, we have de different value for $V_T$ which obviously shifts the obtained curve.

## Slide 56
Folded cascode
The cascode transistor that is added to a transistor in order to increase its output resistance, does  not require to be of the same polarity.
When we introduce a transistor of another polarity, a correct settling of the DC bias current is needed.
This requires an additional transistor.


## Slide 57
Typical folded cascode circuit
A voltage change at the input causes a current change DI = gm1Dvin
This DI causes a DV*
As a consequence of this DV* we get:
–(gm2 +1/Ro2)Dv* as current through  T2 and
 Dv* /Ro3 as current through T3
The current change through T2 I observed as a voltage change at the output


### Notes — Slide 57

Please note that only a limited current change through T3 is observed when compared to the current change through T2. As a consequence,  the major part of the current change is observed at the output.
We conclude that there is only a limited difference between a regular cascode circuit and a folded cascode circuit in terms of current at the output.

## Slide 69
Amplifiers with high $\mathrm{GBW}$
In order to obtain an amplifier with a high $\mathrm{GBW}$ we can have one pole below the $\mathrm{GBW}$ frequency , but no second pole, because otherwise, no sufficient phase margin is obtained.
How can we obtain a good guess of all other poles and zeroes?
A good starting point is link all poles with the capacitance of a node in the circuit.
Example: an RC network


### Notes — Slide 69

In this assessment 1/R is the relationship between the current that is driven to a specific node and the voltage change obtained on this node as a consequence of this current driven to it. As soon as a frequency is obtained whereby more current is running towards the capacitor of this node, we have reached the pole frequency.
The relationship between voltage and current is not only defined by resistances or conductances, but also transconductances as in current mirrors or cascode transistors.


## Slide 70
Parasitic pole of a current mirror
This current mirror get an input current that equals gm1 times the input voltage ($V_{in}$)
This current is loaded into the input conductance of transistor T2 as diode (gm2) and by the parasitic capacitance of this node (Ca)
When the current mirror increases the current by a factor B, it also increases the capacitive load by a factor B


## Slide 71
Consequences of the parasitic pole of the current mirrors
In a single stage opamp, current mirrors enable to increase the $\mathrm{GBW}$ by increasing the factor B of the current mirrors
This method has however important limitations as this method decreases the poles frequencies linked with  these current mirrors such that they get, at a certain point smaller than the $\mathrm{GBW}$
In this case  the phase margin vanishes
