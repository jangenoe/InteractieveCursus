## What is superconductivity?
Most non-magnetic metals and a few ceramic crystals show a resistance that equals exactly zero when the temperature falls below the critical temperature Tc.
Tc strongly depends on:
the current density
the magnetic field
The use of superconductors can be interesting because:
No losses occur in the resistances (dissipation is an important constraint in the scaling down of semiconductor circuits).
There are no longer RC delays. A connection must be modeled by its LC equivalent (ideal transmission line).

## The  birth of a Cooper pair
The  birth of a Cooper pair


### Notes — Slide 4

From the global perspective, a crystal is always neutral. This implies that the charges of the free electrons is compensated by an equal but opposite charge of the ions.  But on a local scale, it is obvious that charges matter and that local potentials are induced. At first glance this yields in a situation that both electron charges will be repelled from each other by their electric field. This will be definitely true if they get close to each other. However, when they reside at a bit longer distance from each other this is not necessary true due to the fact that the first  electron () attracts fixed ion charges from its neighborhood. As a consequence, the surrounding of electron () will look like a positive charge from a distance, and like a negative charge when electron () gets closer. Electron  () will be attracted by electron () and vice versa and they will form a pair that will reside at a fixed distance from each other. This pair is called a Cooper pair, as they have  been discovered by mister Cooper. This Cooper pair formation  is only happening when the movement of the ions, due to temperature is only limited. As a consequence, superconductivity will only occur at low temperatures. The energy of a Cooper pair is lower than the sum of the energies of the two individual electrons. We name this energy difference the  superconductivity gap D.  Cooper pairs can move freely. However, they behave completely different from electrons. They prefer to move together all coordinated. As a consequence, the do not have collisions with the ions of the lattice, and the  movement does not contribute to resistance.

## Josephson junction
A Josephson junction is an utterly thin isolator between  superconductors.
A superconductive current can pass through this junction until a maximum current (i.e. without voltage drop).
As soon as one exceeds this maximum current, a voltage drop D over the junction is obtained (memory behavior)
Removing all current can bring back the superconductivity
