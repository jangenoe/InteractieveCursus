## Emitter follower
The aim of an emitter follower:
no voltage amplification at the output
A higher current at the output with respect to the current at the input(the output has a lower impedance than the input)
In principle, one could get a current amplification b. However, the settling network prevents this.
The AC current from the source, is not only running towards the base of the transistor but also through R1 and R2. Typically, the AC currents are $10\times$ bigger than the base current
These higher current are required because the resistors need to deliver the needed base currents under all circumstances
Emitter follower
For the class B amplifier, we have solved this by the introduction of a current source

## Bootstrapping the emitter follower
The capacitor CB provides a positive feedback $< 1$
The bias currents in R1 and R2 can now be selected such that they need to provide only the DC currents.
The capacitor CB must be rather large

### Notes — Slide 7

This system incorporates two feedback systems:
The positive feedback is the capacitor between input ($V_{in}$) and output. The negative feedback is the connection between the output and the – input.
This system can only be used at low frequencies of short connections. Otherwise, the behavior of the triax is as  a transmission line.
