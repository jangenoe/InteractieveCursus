## Balanced Amplifier
Balanced Amplifier


### Notes — Slide 9

In many case the input impedance of the amplifier does not match the characteristic impedance of the transmission line. So an important reflection of the signal will be obtained on the transmission line at the input.
The same holds  for the output impedance of the amplifier and the transmission line connected to the output.
When two exactly equal amplifiers are available, one would obtain twice the same reflected signal. By adding an  degree phase difference (i.e. +) on one of the signals, both reflected signal will cancel, both at the input and the output.
Please note that in the forward path, signals in phase are added.

### Notes — Slide 10

A transformer hybrid, combined with a transmission line ( e.g. length difference on the PCB) can be used to build a balanced amplifier.
