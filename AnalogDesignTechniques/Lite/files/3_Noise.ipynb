## Combination of noise sources
Noise voltages nor noise currents of different origin can never be added.
Their squares (i.e. power) can be added.
Example: we place R1 and R2 in parallel
By adding the squares of the noise currents we get an equal result as if we would have calculated the noise current of the resistance having an equal value as the parallel connection of both resistances.
Example: we place R1 and R2 in series
By adding the squares of the noise voltages we get an equal result as if we would have calculated the noise voltage of the resistance having an equal value as the series connection of both resistances.
Combination of noise sources

## Noise figure of an amplifier
This noise figure is determined by adding the minimal noise power at the input of the amplifier under matched conditions (Ninput = kToB)
Also in the case that we will never match the amplifier in reality
The noise figure F is defined at To=290°K
The amplification (G) in this equation is the power amplification
Noise figure of an amplifier


### Notes — Slide 9

Why do we select the minimal noise at the input to calculate the noise figure F? If we would add more noise at the input, the relative contribution of the noise of the amplifier would be less, also reducing as a consequence the of the obtained F (i.e. better F). But this would not represent the optimal use of the amplifier.

## Shot noise: origin
When transport of individual electrons through a space charge region occurs, the total current is the sum of individual small currents.
The sum of this current is given by the diode current equation but there is also a variation on this current.
This variation is the shot noise.
Shot noise: origin


## Shot noise of a diode
Shot noise of a diode
Without an external applied voltage the noise is defined by the cut-off current Io that runs in both directions through the diode, yielding a total current zero.
On top of this we add the forward current IDC
This is even a very accurate way to determine the charge q of the electron.

## Validity of this equation
An electron needs a certain time to move through the space charge region: this is the transit time.
This equation is valid until a frequency which is the inverse of the transit time.
This frequency is normally higher than the frequencies in use in our designs.
For higher frequencies, the contribution to this noise decreases.
